```python
site_key = 'Telephone'
type_key = ''

data_params = dict()
data_params["site_tag"] = site_key
data_params["site_name"] = SITE_NAMES[site_key]
data_params["type_tag"] = type_key
data_params['detector_tag'] = 'bd2'
data_params['assembly_type'] = 'kmeans'
data_params['training_set'] = 'all_locations'
file_paths = get_file_paths(data_params)
```

```python
location_classes = pd.read_csv(file_paths['SITE_classes_file'], index_col=0)
(Path(file_paths['SITE_classes_folder']) / 'file_level').mkdir(parents=True, exist_ok=True)
```

```python
def save_individual_file_classes(file_classes):
    print(f'{Path(file_classes.name).stem}')
    save_filename = f'{data_params["detector_tag"]}__{data_params["site_tag"]}_{Path(file_classes.name).stem}_classes'
    file_classes.to_csv(Path(file_paths['SITE_classes_folder']) / f'file_level/{save_filename}.csv')
```

```python
location_classes_grouped_by_file = location_classes.groupby('file_name')
```

```python
location_classes_grouped_by_file.apply(lambda x: save_individual_file_classes(x))
```

```python
location_df = pd.read_csv(Path(file_paths['SITE_folder']) / f"{file_paths['detector_TYPE_SITE_YEAR']}.csv", index_col=0)
(Path(file_paths['SITE_folder']) / 'file_level').mkdir(parents=True, exist_ok=True)
```

```python
def save_individual_file_classified_dets(file_dets):
    print(f'{Path(file_dets.name).stem}')
    save_filename = f'{data_params["detector_tag"]}__{data_params["site_tag"]}_{Path(file_dets.name).stem}_classified'
    file_dets.to_csv(Path(file_paths['SITE_folder']) / f'file_level/{save_filename}.csv')
```

```python
location_classed_dets_grouped_by_file = location_df.groupby('input_file')
```

```python
location_classed_dets_grouped_by_file.apply(lambda x: save_individual_file_classified_dets(x))
```

```python
def save_individual_file_raw_dets(file_dets):
    print(f'{Path(file_dets.name).stem}')
    save_filename = f'{data_params["detector_tag"]}__{data_params["site_tag"]}_{Path(file_dets.name).stem}'
    save_dets = file_dets[file_dets['det_prob']>=0.2].copy()
    save_dets.to_csv(f'/Users/adityakrishna/duty-cycle-analysis/data/raw/{data_params["site_tag"]}/{save_filename}.csv')
```

```python
raw_files = sorted(list(Path(file_paths['raw_SITE_folder']).glob(f'{data_params["detector_tag"]}__{data_params["site_tag"]}*.csv')))
for raw_filepath in raw_files:
    file_dets = pd.read_csv(raw_filepath)
    save_dets = file_dets[file_dets['det_prob']>=0.2].copy()
    save_dets.to_csv(f'/Users/adityakrishna/duty-cycle-analysis/data/raw/{data_params["site_tag"]}/{raw_filepath.name}')
```
